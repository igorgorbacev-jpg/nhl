```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers
```

```python
def remove_outliers_iqr(df, column, multiplier=1.5):
    """
    Удаляет выбросы по методу IQR
    multiplier=1.5 — стандартный, 3.0 — более мягкий
    """
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - multiplier * IQR
    upper_bound = Q3 + multiplier * IQR
    
    # Фильтруем данные
    df_clean = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    
    removed = len(df) - len(df_clean)
    print(f"Удалено {removed} строк ({removed/len(df)*100:.2f}%) из столбца '{column}'")
    print(f"Границы: [{lower_bound:.2f}, {upper_bound:.2f}]")
    
    return df_clean
```

```python
def clean_dataset_iqr(df, columns_to_check, multiplier=1.5):
    """
    Удаляет строки, где есть выбросы хотя бы в одной из указанных колонок
    """
    df_clean = df.copy()
    total_removed = 0
    
    for col in columns_to_check:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - multiplier * IQR
        upper_bound = Q3 + multiplier * IQR
        
        mask = (df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)
        df_clean = df_clean[mask]
        
        removed = len(df) - len(df_clean)
        print(f"{col}: удалено {removed} строк")
    
    print(f"\nВсего удалено: {len(df) - len(df_clean)} строк ({ (len(df) - len(df_clean))/len(df)*100:.2f}%)")
    return df_clean

# Указываем колонки, которые хотим очистить
cols_to_check = ['attendance']
df_clean = clean_dataset_iqr(df, cols_to_check, multiplier=1.5)
df_clean
```

```python
def for_optuna(trial):
    dictt = {}
    learning_rate = trial.suggest_float('learning_rate',0.12,0.14)
    depth = trial.suggest_int('depth',7,9)
    l2_leaf_reg = trial.suggest_int('l2_leaf_reg',5,7)
    min_data_in_leaf = trial.suggest_int('min_data_in_leaf',15,17)
    random_strength = trial.suggest_int('random_strength',1,2)
    iterations = trial.suggest_int('iterations',1000,1001)
    random_seed = trial.suggest_int('random_seed',42,43)
    dictt['learning_rate'] = learning_rate
    dictt['depth'] = depth 
    dictt['l2_leaf_reg'] = l2_leaf_reg
    dictt['iterations'] = iterations
    dictt['random_seed'] = random_seed
    dictt['min_data_in_leaf'] = min_data_in_leaf
    dictt['random_strength'] = random_strength
    new_pipe = Pipeline([
            ('trans_pipe',PowerTransformer(method='yeo-johnson',standardize=False))
        ])
    coln = ColumnTransformer([
            ('id_1',new_pipe,x_train2.columns)
        ],remainder='passthrough')
    pipe_cat = Pipeline([
            ('coln',coln),
            ('cat',CatBoostRegressor(**dictt))
        ])
    pipe_cat.fit(X_train,y_train)
    pred = pipe_cat.predict(X_eval)
    mae = mean_absolute_error(y_eval,pred)
    rmse = np.sqrt(mean_squared_error(y_eval,pred))
    r2 = r2_score(y_eval,pred)
    return r2

    
    
    
```

```python
def optuna_ligh(trial):
    d = {}
    learning_rate = trial.suggest_float('learning_rate',0.11,0.12)
    max_depth = trial.suggest_int('max_depth',12,14)
    num_leaves = trial.suggest_int('num_leaves',140,150)
    reg_alpha = trial.suggest_int('reg_alpha',2,3)
    reg_lambda = trial.suggest_int('reg_lambda',0,1)
    min_split_gain = trial.suggest_float('min_split_gain',0.3,0.4)
    n_estimators = trial.suggest_int('n_estimators',1000,1001)
    random_state = trial.suggest_int('random_state',42,43)
    d['learning_rate'] = learning_rate
    d['max_depth'] = max_depth
    d['num_leaves'] = num_leaves
    d['n_estimators'] = n_estimators
    d['random_state']= random_state
    d['reg_alpha'] = reg_alpha
    d['reg_lambda'] = reg_lambda
    d['min_split_gain'] = min_split_gain
    new_pipe = Pipeline([
                ('trans_pipe',PowerTransformer(method='yeo-johnson',standardize=False))
            ])
    coln = ColumnTransformer([
                ('id_1',new_pipe,x_train2.columns)
            ],remainder='passthrough')
    pipe_lgbm = Pipeline([
                ('coln',coln),
                ('lgbm',LGBMRegressor(**d))
            ])
    pipe_lgbm .fit(X_train,y_train)
    predict_for_lgbm = pipe_lgbm .predict(X_eval)
    r2_for_lgbm = r2_score(y_eval,predict_for_lgbm)
    return r2_for_lgbm
```

```python
class Oneorord(BaseEstimator, TransformerMixin):
    def __init__(self,threshold=4):
        self.threshold = threshold
    def fit(self,X,y=None):
        self.categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
        self.for_one = []
        self.for_order = []
        for x_trans in self.categorical_cols:
            if X[x_trans].nunique() < self.threshold:
                self.for_one.append(x_trans)
            else:
                self.for_order.append(x_trans)
        self.one = OneHotEncoder(handle_unknown="ignore",sparse_output=False)
        self.order = OrdinalEncoder(handle_unknown='use_encoded_value',unknown_value=10000)
        if self.for_one:
            self.one.fit(X[self.for_one])
        if self.for_order:
            self.order.fit(X[self.for_order])
        return self
    def transform(self,X):
        new_x = X.copy()
        cols_to_drop = self.for_one + self.for_order
        if cols_to_drop:
            new_x = new_x.drop(columns=cols_to_drop, errors='ignore')
        if self.for_one:
            one_hot_data = self.one.transform(X[self.for_one])
            one_hot_df = pd.DataFrame(
                one_hot_data,
                columns=self.for_one,
                index=X.index
        )
            new_x = pd.concat([new_x, one_hot_df], axis=1)
        if self.for_order:
            order_hot_data = self.order.transform(X[self.for_order])
            ordinal_df = pd.DataFrame(
                    order_hot_data,
                    columns=self.for_order,
                    index=X.index
            )
            new_x = pd.concat([new_x, ordinal_df], axis=1)
        return new_x
        
        
```
